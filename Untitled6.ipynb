# 3. Data Imputation & Cleaning

### Step 14: Restore and Clean Customer Names
Since the `customer_name` column was accidentally overwritten with nulls, we restore the original names from our raw file matching on index, and safely strip whitespace.

```python
raw_df = pd.read_csv('dirty_sales_data_2000x30.csv')
df['customer_name'] = raw_df.loc[df.index, 'customer_name'].astype(str).str.strip()
df['customer_name'] = df['customer_name'].replace('nan', None)
display(df[['customer_name']].head())
```

### Step 15: Clean Shipping Cost and Satisfaction Score
We remove special characters (currency symbols, commas, and trailing unit text) from `shipping_cost` and `satisfaction_score` and convert them to numeric values.

```python
# Clean shipping_cost
df['shipping_cost'] = df['shipping_cost'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False).str.strip()
df['shipping_cost'] = pd.to_numeric(df['shipping_cost'], errors='coerce')

# Clean satisfaction_score
df['satisfaction_score'] = df['satisfaction_score'].astype(str).str.replace(' units', '', regex=False).str.strip()
df['satisfaction_score'] = pd.to_numeric(df['satisfaction_score'], errors='coerce')

display(df[['shipping_cost', 'satisfaction_score']].head())
```
