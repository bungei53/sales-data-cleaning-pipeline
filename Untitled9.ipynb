```python
category_sammary=df.groupby('Category').agg(
    products=('Product_ID','count'),
    revenue=('Final_Price(Rs.)','sum'),
    average_price=('Final_Price(Rs.)','mean'),
    min_price=('Final_Price(Rs.)','min'),
    max_price=('Final_Price(Rs.)','max'),
    median_price=('Final_Price(Rs.)','median')
).sort_values(['products','revenue','average_price','min_price','max_price','median_price'],ascending=False)
```

```python
print(category_sammary)
```
